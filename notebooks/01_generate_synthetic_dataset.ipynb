# Генерация синтетического датасета
LLM agent-core создаёт вопросы и короткие дословные свидетельства. Генератор проверяет цитаты и смещения по исходному корпусу.

```python
from pathlib import Path
import sys

root = Path.cwd() if (Path.cwd() / "config.py").exists() else Path.cwd().parent
sys.path.insert(0, str(root))
from evalkit.synthetic import generate_synthetic
```

```python
record = generate_synthetic(
    size=20, on_event=lambda message, progress=None: print(message)
)
record
```

```python
import json
import pandas as pd

qa = [
    json.loads(line)
    for line in Path(record["dataset_config"]["qa_path"]).read_text().splitlines()
]
pd.DataFrame(qa)[["question_type", "question", "reference_answer", "gold_spans"]].head(
    10
)
```

Набор уже доступен в GUI и CLI. Для прогона используйте `python -m evalkit.cli run -d <dataset_id>`. Метаданные содержат модель, seed, хэши файлов и расход токенов генерации.
